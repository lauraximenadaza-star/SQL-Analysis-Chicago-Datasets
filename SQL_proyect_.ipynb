# **Analysis**

**Total number of crimes recorded**

```python
%%sql
SELECT COUNT(*) AS total_crimes
FROM chicago_crime_data;
```

**Community areas with per capita income < 11,000 USD**

```python
%%sql
SELECT community_area_name,
       community_area_number,
       per_capita_income
FROM   census_data
WHERE  per_capita_income < 11000
ORDER  BY per_capita_income;
```

The data shows that the communities analyzed have relatively low per capita income levels. Riverdale has the lowest per capita income at 8,201, while West Garfield Park has the highest at 10,934. The difference between the areas is small, suggesting similar economic conditions and potentially shared socioeconomic challenges in these Chicago communities.

**Case numbers for crimes involving minors**

```python
%%sql
SELECT case_number,
       primary_type,
       description,
       date
FROM   chicago_crime_data
WHERE  description LIKE '%MINOR%'
ORDER  BY date;
```

Records show cases related to violations of Chicago's liquor laws. Both incidents involve minors: one for illegal consumption of alcohol and the other for selling or delivering alcohol to a minor. The cases occurred in 2004 and 2005, demonstrating enforcement and reporting of alcohol regulations in the city during that period.

**Kidnappings involving children**

```python
%%sql
SELECT case_number,
       date,
       description,
       arrest,
       community_area_number
FROM   chicago_crime_data
WHERE  primary_type = 'KIDNAPPING'
  AND  description  LIKE '%CHILD%'
ORDER  BY date;
```

The record pertains to a case of child abduction by an unknown person reported in 2007 in Chicago's Community Area 25. The value ARREST = 0 indicates that no arrests were made in connection with the incident at that time. This type of crime is considered highly serious due to the risk to minors and the potential for immediate resolution.

**Types of crime recorded in schools (without repetitions)**

```python
%%sql
SELECT DISTINCT primary_type
FROM   chicago_crime_data
WHERE  location_description LIKE '%SCHOOL%'
ORDER  BY primary_type;
```

The data reveals some of the most frequent types of crimes reported in Chicago. Crimes involving physical violence, such as assault and battery, predominate, along with criminal damage and trespass. Drug-related offenses and public peace violations also appear, reflecting a combination of security, community relations, and substance abuse issues in various parts of the city.

**The 5 areas with the highest percentage of households below the poverty line.**

```python
%%sql
SELECT   community_area_name,
         percent_households_below_poverty,
         hardship_index
FROM     census_data
ORDER BY percent_households_below_poverty DESC
LIMIT    5;
```

The data reveals high levels of poverty and socioeconomic hardship in several Chicago communities. Riverdale presents the most critical situation, with 56.5% of households below the poverty line and a hardship index of 98, followed by Fuller Park and Englewood. These figures suggest significant economic and social challenges, including reduced access to resources, employment, and development opportunities in these areas.

**Community area with the highest number of crimes**

```python
%%sql
SELECT   community_area_number,
         COUNT(*) AS total_crimes
FROM     chicago_crime_data
WHERE    community_area_number IS NOT NULL
GROUP BY community_area_number
ORDER BY total_crimes DESC
LIMIT    1;
```

The results show that Chicago's Community Area 25 recorded a total of 43 crimes during the analyzed period. This indicates significant criminal activity in that community and can serve as a basis for comparing crime levels among different areas of the city.

**Community area with the highest difficulty index (subconsultation)**

```python
%%sql
SELECT community_area_name,
       hardship_index
FROM   census_data
WHERE  hardship_index = (SELECT MAX(hardship_index) FROM census_data);
```

Data indicates that Riverdale has a hardship index of 98, one of the highest possible levels. This reflects very difficult socioeconomic conditions, generally associated with high levels of poverty, unemployment, and limited access to resources and opportunities within Chicago.

**Name of the most criminalized community area (subconsultation)**

```python
%%sql
SELECT community_area_name,
       community_area_number
FROM   census_data
WHERE  community_area_number = (
           SELECT community_area_number
           FROM   chicago_crime_data
           WHERE  community_area_number IS NOT NULL
           GROUP  BY community_area_number
           ORDER  BY COUNT(*) DESC
           LIMIT  1)
```

The results show that Austin, corresponding to community area 25, is the area with the highest number of reported crimes in the Chicago analysis. This suggests a high concentration of criminal activity in this community compared to other areas of the city.

**Crime and socioeconomic context by community area**

```python
%%sql
SELECT   cd.community_area_name,
         cd.percent_households_below_poverty,
         cd.per_capita_income,
         cd.hardship_index,
         COUNT(cr.id)                    AS total_crimes,
         ROUND(COUNT(cr.id) * 1.0 /
               NULLIF(cd.percent_households_below_poverty, 0), 2) AS crimes_per_poverty_pct
FROM     census_data            cd
LEFT JOIN chicago_crime_data    cr
       ON cd.community_area_number = cr.community_area_number
GROUP BY cd.community_area_name,
         cd.percent_households_below_poverty,
         cd.per_capita_income,
         cd.hardship_index
ORDER BY total_crimes DESC
LIMIT 15;
```

The analysis shows a partial relationship between poverty, hardship, and crime in Chicago communities. Austin leads in total number of crimes (43) and has a high hardship score (73), suggesting significant socioeconomic pressure. Communities like Englewood and North Lawndale also combine high levels of poverty and hardship with relevant crime.

However, higher-income areas like Near North Side and Lake View still register significant relative crime rates, indicating that crime is not solely dependent on poverty. Urban factors, population density, and economic activity can also influence crime patterns.

**Top 10 areas by crime with their difficulty index**

```python
%%sql
WITH crime_counts AS (
    SELECT   community_area_number,
             COUNT(*) AS total_crimes
    FROM     chicago_crime_data
    WHERE    community_area_number IS NOT NULL
    GROUP BY community_area_number
),
ranked AS (
    SELECT   cc.community_area_number,
             cc.total_crimes,
             cd.community_area_name,
             cd.hardship_index,
             cd.percent_households_below_poverty
    FROM     crime_counts cc
    JOIN     census_data  cd
          ON cc.community_area_number = cd.community_area_number
    ORDER BY cc.total_crimes DESC
    LIMIT 10
)
SELECT * FROM ranked;
```

The data shows that several communities with high levels of poverty and hardship also have high crime rates in Chicago. Austin tops the list with 43 crimes, accompanied by a hardship of 73 and 28.6% of households living in poverty. Similarly, Englewood and West Englewood exhibit both high poverty and very high hardship.

However, there are exceptions, such as the Near North Side, which has low hardship and more favorable incomes, but still registers a significant number of crimes. This indicates that crime is influenced by multiple factors beyond economic conditions, including urban density, commercial activity, and the social dynamics of each community.

**Arrest rate by type of crime**

```python
%%sql
SELECT   primary_type,
         COUNT(*)                                   AS total_incidents,
         SUM(CASE WHEN arrest = 1 THEN 1 ELSE 0 END) AS arrests,
         ROUND(
             SUM(CASE WHEN arrest = 1 THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
         2) AS arrest_rate_pct
FROM     chicago_crime_data
GROUP BY primary_type
HAVING   COUNT(*) > 5
ORDER BY arrest_rate_pct DESC
LIMIT    15;
```

The data shows significant differences in arrest rates depending on the type of crime in Chicago. Crimes such as prostitution and narcotics have very high arrest rates, close to or equal to 100%, suggesting greater ease of detection and apprehension in police operations.

In contrast, crimes such as burglary, motor vehicle theft, and theft have very low arrest rates, indicating greater difficulty in identifying perpetrators. Although battery and assault are frequent, their arrest rates are also relatively low, which may reflect investigative limitations or complexity in these cases.

**Conclusions**

The following findings can be drawn from the SQL analysis performed on the three Chicago datasets:

Crime concentration: A few community areas account for the majority of reported crimes, and these areas coincide with those exhibiting the highest hardship index and the highest percentage of households below the poverty line.
School crimes: Several types of crimes were recorded in or near educational institutions, indicating a localized school safety problem.
School safety: There is variability in the safety score across school types (elementary, middle, high school), and some schools in high-crime areas maintain relatively good scores.
Differential arrest rates: Certain types of crime—such as narcotics offenses—have markedly different arrest rates than property crimes, which may reflect both police pursuit patterns and evidentiary difficulties.

Value of JOINs: Crossing the three tables allows contextualizing crime data with socioeconomic indicators, which enhances the explanatory power of the analysis compared to isolated queries on a single table.
